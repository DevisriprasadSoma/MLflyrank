## 1. My lane as an ML task (type)



### Lane 2 — Refresh / Content Opportunity Scoring

**ML task type: Ranking**

I frame this lane as a ranking problem because the practical decision is which pages should be reviewed first. The system would use observable page-level signals to produce a review-priority ranking rather than making an automatic decision about what should happen to a page.

The output would help a content or SEO reviewer decide where to spend limited review time first. The possible actions after human review could include refreshing, expanding, protecting, pruning, or monitoring a page.

## 2. Target or proxy



### Target / Proxy

For the starter exploration, I would use `trend_direction` as a proxy for whether a page is currently showing a downward trend.

The proxy can be represented as:

- `1` = current trend is down
- `0` = current trend is not down

This is useful for learning the framing, but it is not a perfect future outcome. A stronger final target would ideally measure a future change after a page is reviewed or after a defined time window.

The model should use signals that are available before the decision, rather than features that directly reveal the outcome.

## 3. Success metric



### Success Metric

I would use **Precision@K** as the main ranking metric, with Precision@50 as an initial example.

Precision@50 asks: among the 50 pages ranked highest for review, what fraction are actually classified as declining by the starter proxy?

This metric matches the decision because review capacity is limited. If a reviewer can only inspect a fixed number of pages, the quality of the top part of the ranking matters more than the average performance across every page.

I would also compare the ML ranking against a simple hand-written baseline so that improvement is measured against something understandable rather than against no baseline.

## 4. The unit of analysis, as a real dataframe



```python
import os
import pandas as pd

repo_path = "/content/flyrank-ml-internship-starter"
csv_path = f"{repo_path}/data/raw/content_refresh_anonymized.csv"

if not os.path.exists(csv_path):
    !git clone https://github.com/flyrank-bih/flyrank-ml-internship-starter.git

df = pd.read_csv(csv_path)

print("Unit of analysis: one row = one content page")
print("Shape:", df.shape)

df[[
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count",
    "trend_direction"
]].head(10)
```

### Unit of analysis

The unit of analysis is **one content page**. Each row represents one page, and the columns contain observable page-level signals such as content age, freshness, impressions, average position, CTR, and word count.

The target/proxy is page-level as well, so the inputs and the decision operate at the same unit of analysis.

```python
df["declining_proxy"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

df[[
    "trend_direction",
    "declining_proxy"
]].head(10)
```

## 5. Why ML beats a fixed rule here



### Why ML could beat a fixed rule

A fixed rule such as "review pages that are old and have high impressions" is easy to understand, but it requires manually choosing thresholds and combinations of signals.

ML could be useful if different signals interact in ways that are difficult to capture with one fixed rule. It can learn patterns from multiple observable features and produce a ranking that can be compared against a simple baseline.

However, ML is not automatically better. I would only prefer the ML approach if it improves the chosen ranking metric on a suitable validation setup and provides useful, actionable rankings. A simple rule should remain the baseline.
